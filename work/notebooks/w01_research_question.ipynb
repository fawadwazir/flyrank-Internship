## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision improved:** which pages enter this cycle's content-refresh review queue, given
a reviewer can only look at a limited number of pages per cycle.

**Who acts on it:** a content reviewer / SEO strategist with limited review capacity.

**Action taken:** the reviewer opens each flagged page, checks the reason codes, and decides
to refresh, expand, protect, or leave it alone.

**Cost of a wrong call:**
- False positive: reviewer time is spent on a page that wasn't actually a problem — real
  cost is opportunity cost, since that time could have gone to a page that needed it.
- False negative: a genuinely declining, high-demand page keeps losing visibility unnoticed
  until recovering it is harder and more expensive.

Because the cost is asymmetric per situation (wasted time vs. compounding traffic loss),
this is a ranking/prioritization problem, not a strict classification problem — Precision@K
matches how the output actually gets used.
